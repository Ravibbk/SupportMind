# SupportMind: Exploratory Data Analysis (EDA)
**Author:** SupportMind ML Engineering Team  
**Dataset:** NovaCart Customer Tickets (1,500 rows) & Fulfillment Orders (500 rows)

## Objectives
1. Ingest and profile synthetic e-commerce customer support messages and orders.
2. Inspect data hygiene: nulls, duplicates, and noisy character encodings.
3. Analyze label distributions across support categories (`billing`, `delivery`, `refund`, `account`, `product`, `other`).
4. Profile customer text characteristics: character/word length distributions and TF-IDF key terms.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

tickets_path = Path('../backend/data/tickets_train.csv')
orders_path = Path('../backend/data/orders.csv')

df_tickets = pd.read_csv(tickets_path)
df_orders = pd.read_csv(orders_path)
print(f"Loaded {len(df_tickets)} tickets and {len(df_orders)} orders.")
```

### 1. Data Cleaning & Hygiene Checks

```python
# Check missing values
print("Missing values in tickets:")
print(df_tickets.isnull().sum())

# Check duplicates
print("\nDuplicate ticket texts:", df_tickets.duplicated(subset=['text']).sum())
print("\nSample ticket rows:")
df_tickets.head(5)
```

### 2. Category Distribution & Class Balance
Understanding class imbalance is critical to choosing the correct evaluation metric (e.g. Stratified K-Fold and Macro F1 rather than raw accuracy).

```python
cat_counts = df_tickets['category'].value_counts()
print("Class balance percentages:")
print((cat_counts / len(df_tickets) * 100).round(2))

plt.figure(figsize=(8, 4))
cat_counts.plot(kind='bar', color='#3b82f6', edgecolor='#1e293b')
plt.title('Support Ticket Category Counts')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

### 3. Message Length Distribution (Tokens and Characters)

```python
df_tickets['word_count'] = df_tickets['text'].str.split().str.len()
df_tickets['char_count'] = df_tickets['text'].str.len()

print("Word count summary statistics:")
print(df_tickets['word_count'].describe())

fig, ax = plt.subplots(figsize=(8, 4))
df_tickets['word_count'].hist(bins=25, ax=ax, color='#6366f1', edgecolor='#1e293b')
ax.set_title('Distribution of Customer Message Word Counts')
ax.set_xlabel('Word Count')
ax.set_ylabel('Number of Inquiries')
plt.show()
```

### 4. Orders Fulfillment Analysis

```python
print("Order Status Breakdown:")
print(df_orders['status'].value_counts())
print("\nAverage order value: $", round(df_orders['price_usd'].mean(), 2))

plt.figure(figsize=(7, 4))
df_orders['status'].value_counts().plot(kind='bar', color='#10b981', edgecolor='#1e293b')
plt.title('NovaCart Order Status Breakdown (500 Orders)')
plt.xlabel('Order Status')
plt.ylabel('Count')
plt.show()
```
